# Synthetic generation

This notebook reads cached generator metadata. It does not fit CTGAN or the Gaussian copula.

```python
from pathlib import Path
import sys
import pandas as pd
from IPython.display import display, Image, Markdown

ROOT = Path.cwd().resolve()
if not (ROOT / "pyproject.toml").is_file():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / "src"))

from generators.tabular import generation_dir
import json
base = ROOT / "data" / "synthetic" / "main"
rows = []
for path in sorted(base.glob("*/*/generation.json")):
    payload = json.loads(path.read_text())
    rows.append({
        "generator": payload.get("generator"),
        "seed": payload.get("seed"),
        "status": payload.get("status"),
        "rows": payload.get("n_rows"),
        "seconds": payload.get("fit_seconds"),
        "invalid_target_rows": payload.get("invalid_target_rows"),
        "accelerator": payload.get("accelerator"),
    })
frame = pd.DataFrame(rows)
display(frame)
print("test metrics are not computed in this notebook")
```
